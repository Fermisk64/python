```python
# qualifying for an eye exam
def qualified_for_exam (age, qualifying_condition, time_since_last_exam):
    '''assume (1) age is passed as an int for the age of the person, 
    (2) qualifying_condition is passed as boolean indicating whether qualified for an exam 
    (3) time_since_last exam is passed as an int for the number of months since last exam'''
    
    if 0< age <= 19:
        if time_since_last_exam >= 12:
            return "You are eligible for one eye exam and any minor assessment"
        else:
            return "It has not been more than 12 months for another eye exam"
    elif 20 <= age <= 64:
        if qualifying_condition:   #Since it isn't defined, it automatically considers it as true
            if time_since_last_exam >= 12:
                return "You are eligible for one major eye exam and 2 additional follow-up minor assessment"
            else:
                return "It hasn't been 12 months since your last eye exam"
        else:
            return "You do not have a qualifying eligibla medical condition"
    elif age >= 65:
        if time_since_last_exam >= 18:
            return "You are eligible for one major eye exam or two minor assessments"
        else:
            return "It hasn't been 18 months since your last exam"
    else:
        return "Invalid age input"
```

```python
# class: blueprint or template for creating objects
# define the protperties (attributes) and behaviours (methods) that objects of that class will
# have user_defined datatypes

# object: specific instances of a class
# have properties (attributes)  and behaviours (methods)

# method: funtion that is associated with a calss and can be called on objects of that class
# behaviour of the objects

# attribute: pieces of data that belong to an object
# represent the state of the object


kcar_name = 'ketchup'   # these are all attributes
kcar_make = 'Volkswagen'
kcar_model = 'UP!'
kcar_colour = 'red'

dcar_name = 'Taylor'
dcar_make = 'Honda'
dcar_model = 'Civic'
dcar_colour = 'Black'

# we can define these attributes and more for anyones' car but it is cumbersome
# to avoide the time and make it simple, we define a class

# __init__ : 

class Car:
    def __init__(self, name, make, model, colour):
        self.name = name # methods
        self.make = make
        self.model = model
        self.colour = colour

    def honk(self):
        print('beep beep')
```

```python
# operations in multiple dimentions

tens = np.arange(0, 120, 10).reshape(3, 4) # array from 0 upto 120 in 10 increments; then resape into 3x4 array
tens
```

```python
horizontal = np.array([-5, -10, -15, -20])
horizontal
```

```python
tens + horizontal # adds each element of horizontal that is 1D to each row of tens in order
```

```python
vertical = np.array([[100], 
                    [200], 
                    [300]]) # a 3x1 vector
vertical
```

```python
tens + vertical # adds each column of vertical to each column of tens vector in order of top to bottom
```

```python
# we can apply methods to multidimensional arrays but we need to add axis
# axis = 0: column-wise opperation (going down the rows)
# axis = 1: row-wise operation (across the column)

tens
```

```python
tens.mean(axis=0) # 0 means columnwise. for each column, calculates the mean and returns it in a 1D
```

```python
tens.mean(axis=1)
```
